1. Macro: Exponential Expansion and Pearl Control
Your early game must focus entirely on multiplying your action economy (number of dragons) and mapping the board.

Split Aggressively Early: A single dragon can only consume one pearl or cover one tile per turn. Splitting early allows you to cover more ground, increase your vision radius, and gather pearls exponentially faster. Set a conservative "safe split" threshold (e.g., splitting when length reaches 6 to create a length 3 child) so both parent and child have a buffer against accidental death.

Determine Map Symmetry: The map is guaranteed to be symmetric (horizontal, vertical, or 180-degree), but you aren't told which. Program your early scouts to observe the map layout and compare it to their spawn coordinates. Once you calculate the symmetry line, you instantly know the rough location of the enemy team and the mirrored locations of all pearls you discover.

Global Pearl Timer Tracking: Since pearls spawn on deterministic timers [min_gap, max_gap], your dragons should memorize the exact location and countdown of every tile they see. Build a centralized array or priority queue in your code to route your dragons to tiles exactly as their countdowns hit zero.

2. Micro: Combat, Trapping, and Sprinting
Combat in this game is not about raw damage; it is about geometry, positioning, and forcing the enemy to crash.

The Sprinting ROI (Return on Investment): Sprinting is your most powerful combat tool, but it costs your most valuable resource (length). You should strictly program your dragons to only sprint if the action has a guaranteed positive return.

Good Sprint: Sprinting 2 extra tiles (costs 2 length) to cut off and kill an enemy dragon of length 10 (drops 5 pearls).

Bad Sprint: Sprinting to beat an enemy to a single pearl (costs 1 length to gain 1 pearl, resulting in a net-zero gain but risking positioning).

Body Trapping and Wrap-Around: Because the map wraps around (toroidal topology), you cannot rely on map edges to trap enemies. You must use Kelp formations and your own body segments to create "boxes" that the enemy's head cannot escape from.

Head-to-Head Avoidance: If your dragon is longer than the enemy, never take a head-to-head collision. A head-to-head kills both dragons. If you have the length advantage, your goal is to survive and win the tie-breaker; force the smaller enemy into your body instead.

3. Scouting and Risk Management
Vision and information asymmetry will win you the late game.

Portal Scouting (The "Canary" Strategy): Portals are incredibly dangerous because you cannot see through them, and an enemy or Kelp could be waiting directly on the other side. Do not send your longest, most valuable dragons through a portal blindly. Instead, split off a minimum-length child (length 2) to act as a scout, send it through the portal, and use it to secure the landing zone.

Encrypted Sonar: Sonar travels in a straight line and stops at the first dragon it hits—which means enemies can intercept your messages. Because you can only send a 32-bit integer, use bitwise operations to encrypt your messages. For example, XOR the message with a shared secret key (hardcoded into your bot) so that if an enemy intercepts your Sonar, they cannot decode your target coordinates or pearl spawn timers.

Cannibalization: If one of your dragons is hopelessly trapped (by Kelp or enemies) and is going to die, have it sprint into a wall on purpose on your terms. This instantly drops its pearls, allowing a nearby teammate to swoop in and collect them before the enemy does.
